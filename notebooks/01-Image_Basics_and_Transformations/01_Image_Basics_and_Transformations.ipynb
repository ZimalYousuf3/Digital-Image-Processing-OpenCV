# **Image Basics and Transformations using OpenCV**

**Objective:** Learn the basic operations of Digital Image Processing using Python, OpenCV and NumPy.

**Tasks covered:**

2. Checking image shape
3. Grayscale conversion
4. Saving an image
5. Average brightness
6. Thresholding
7. Resizing
8. Flipping
9. Drawing a rectangle

**Tools:** Python, OpenCV, NumPy, Google Colab

```python
# ------------------------------------------------------------
# Step 1: Install and import required libraries
# cv2    : OpenCV, used for image processing
# numpy  : used for mathematical operations on image arrays
# cv2_imshow : Colab-friendly function to display images
# files  : used to upload files from the local computer
# ------------------------------------------------------------

!pip install -q opencv-python
import cv2, numpy as np
from google.colab.patches import cv2_imshow
from google.colab import files
```

```python
# ------------------------------------------------------------
# Step 2: Upload and read the image
# files.upload() lets us choose an image from our computer.
# cv2.imread() reads the image as a NumPy array in BGR format.
# ------------------------------------------------------------

# Uploading image

uploaded = files.upload()
fname = list(uploaded.keys())[0]

# Reading image
img = cv2.imread(fname)
print("Loaded: ", fname)
```

```python
# ------------------------------------------------------------
# Step 3: Check the image dimensions
# img.shape returns (Height, Width, Channels)
# Channels = 3 means a color image (Blue, Green, Red)
# ------------------------------------------------------------

# Shape of image (Height, Width, Height)

print(f"Shape (H, W, C): {img.shape}")
```

```python
# ------------------------------------------------------------
# Step 4: Display the original image
# ------------------------------------------------------------

cv2_imshow(img)
```

```python
# ------------------------------------------------------------
# Step 5: Convert the image to grayscale
# A grayscale image has only 1 channel.
# Pixel values range from 0 (black) to 255 (white).
# ------------------------------------------------------------

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
# ------------------------------------------------------------
# Step 6: Save the grayscale image as a new file
# cv2.imwrite(filename, image)
# ------------------------------------------------------------

cv2.imwrite("gray.jpg", gray)
print("Saved gray.jpg")
```

```python
# ------------------------------------------------------------
# Step 7: Calculate the average brightness
# np.mean() gives the average of all pixel values.
# Close to 0 = dark image, close to 255 = bright image.
# ------------------------------------------------------------

avg = float(np.mean(gray))
print(f"Average brighness level: {avg:.2f}")  # round(avg, 2)
```

```python
# ------------------------------------------------------------
# Step 8: Apply binary thresholding
# Pixels greater than 128 become 255 (white).
# Pixels less than or equal to 128 become 0 (black).
# Parameters: image, threshold value, max value, threshold type
# ------------------------------------------------------------

_, bw = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY)
cv2_imshow(bw)
```

```python
# ------------------------------------------------------------
# Step 9: Resize the image to half of its original size
# gray.shape gives (height, width)
# cv2.resize() needs size as (width, height)
# // is integer division because pixel sizes must be whole numbers
# ------------------------------------------------------------

h, w = gray.shape
small = (cv2.resize(gray, (w//2, h//2)))
cv2_imshow(small)
```

```python
# ------------------------------------------------------------
# Step 10: Flip the image
# Flip code:  1 = horizontal,  0 = vertical,  -1 = both
# ------------------------------------------------------------

flip = cv2.flip(img, 1)
cv2_imshow(flip)
```

```python
# ------------------------------------------------------------
# Step 11: Draw a red rectangle on the image
# img.copy() keeps the original image unchanged.
# Parameters: image, top-left (x, y), bottom-right (x, y),
#             color in BGR (0, 0, 255 = red), thickness
# ------------------------------------------------------------

rect_img = img.copy()
cv2.rectangle(rect_img, (40, 40), (300, 300), (0, 0, 255), 3)
cv2_imshow(rect_img)
```
